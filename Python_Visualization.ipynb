```python
# X-axis is for time
year = df['Years']

# Y-axis is for our different basket items
all_items = df['All-items']
food = df['Food']
shelter = df['Shelter']

#plot the time series of given dataset
plt.plot(year, all_items)
plt.plot(year, food, color='orange')
plt.plot(year, shelter, color='green')

# title for the graph
plt.title('CPI Basket Items by Year')

# to make it easier to read, rotate the x-axis tick labels at 30degree and towards right
plt.xticks(rotation=30, ha='right')

# add x and y labels
plt.xlabel('Year')
plt.ylabel('All Items')
plt.show()
```

```python
import seaborn as sns

# seaborn has a number of elegant built-in themes
sns.set_style("dark")

# writing code for a line chart is nearly as easy as in basic pandas
sns.lineplot(x = 'Years', y = 'All-items', data = df)
sns.lineplot(x = 'Years', y = 'Food', data = df, color='orange')
sns.lineplot(x = 'Years', y = 'Shelter', data = df, color='green')
plt.title("CPI 2002-2024");
```

```python
import bokeh
from bokeh.models import ColumnDataSource, DatetimeTickFormatter, FixedTicker
from bokeh.plotting import figure, show
from bokeh.io import output_notebook

# Convert Years into datetime
df['YearDate'] = pd.to_datetime(df['Years'].astype(int), format='%Y')

# Bringing in data from pandas is relatively easy
source = ColumnDataSource(df)

# Bokeh's syntax, like Seaborn's, is easy to write and understand
p = figure(x_axis_type="datetime", width=800, height=350, title="CPI 2002-2024")

# Plot the three lines
p.line('YearDate', 'All-items', source=source, color = 'blue')
p.line('YearDate', 'Food', source=source, color='orange')
p.line('YearDate', 'Shelter', source=source, color='green')

# Format x-axis to show only years
p.xaxis.formatter = DatetimeTickFormatter(years="%Y")

# Rotate labels for readability
p.xaxis.major_label_orientation = 1.0 # ~90 degrees

# we can easily output our line chart as an .html file
#output_file("ts.html")
output_notebook()
show(p)
```

```python

categories = [c for c in df.columns if c != 'Years']

short_labels = {c: c for c in categories}  # adjust if your column names are long

fig, ax = plt.subplots(figsize=(12, 6))

for cat in categories:
    lw = 2.5 if cat in ['All-items', 'Shelter'] else 1.4
    ls = '--' if cat not in ['All-items', 'Food', 'Shelter', 'Transportation'] else '-'
    ax.plot(df['Years'], df[cat], linewidth=lw, linestyle=ls, label=cat)

ax.axvspan(df['Years'].max() - 5, df['Years'].max(), alpha=0.07, color='steelblue')
ax.axvline(df['Years'].max() - 5, color='gray', linestyle=':', linewidth=1, alpha=0.7)
ax.axvline(df['Years'].max(), color='gray', linestyle=':', linewidth=1, alpha=0.7)

ax.set_title('Canadian CPI by Category (2002–2022)', fontsize=14, fontweight='bold')
ax.set_xlabel('Year')
ax.set_ylabel('CPI Index (2002 = 100)')
ax.legend(loc='upper left', fontsize=7.5, framealpha=0.9)
ax.grid(axis='y', alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
```

```python
categories = [c for c in df.columns if c not in ['Years', 'YearDate']]

year_end   = df['Years'].max()
year_start = year_end - 5

start = df[df['Years'] == year_start][categories].values[0]
end   = df[df['Years'] == year_end][categories].values[0]
growth = ((end - start) / start * 100).round(1)

sorted_items = sorted(zip(categories, growth), key=lambda x: x[1])
labels = [c for c, _ in sorted_items]
values = [v for _, v in sorted_items]

fig, ax = plt.subplots(figsize=(10, 5))
bars = ax.barh(labels, values,
               color=['#e74c3c' if v > 0 else '#3498db' for v in values],
               edgecolor='white', height=0.65)

for bar, val in zip(bars, values):
    ax.text(val + 0.2, bar.get_y() + bar.get_height() / 2,
            f'+{val:.1f}%' if val > 0 else f'{val:.1f}%',
            va='center', fontsize=9)

ax.axvline(0, color='black', linewidth=0.8)
ax.set_title(f'CPI Growth by Category: {year_start}–{year_end}', fontsize=13, fontweight='bold')
ax.set_xlabel('Percentage Change (%)')
ax.grid(axis='x', alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
```
